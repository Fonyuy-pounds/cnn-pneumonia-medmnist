# 🟡 LEVEL 4 — CNN BASIC QUESTIONS

---

### Q67: What is a CNN?

A neural network designed for grid-like data (images) that uses convolutional layers with learnable filters to extract spatial features.

---

### Q68: Why are CNNs good for images?

Local connectivity, parameter sharing, translation invariance, hierarchical learning, and spatial awareness.

---

### Q69: What is a convolution?

A mathematical operation where a filter slides over the input, computing element-wise multiplication and summation at each position.

---

### Q70: What is a filter/kernel?

A small matrix of learnable weights that slides over the input to detect specific patterns.

---

### Q71: What does a filter learn?

A specific visual pattern: edges, textures, corners, or complex features depending on depth.

---

### Q72: What is a feature map?

The output of applying a filter to the input — shows where the filter's pattern was detected.

---

### Q73: What is the difference between an image and a feature map?

| Aspect | Image | Feature Map |
|--------|-------|-------------|
| Content | Raw pixels | Detected patterns |
| Channels | 1 or 3 | Number of filters |
| Values | 0-255 | Real numbers |

---

### Q74: What is a kernel size?

The height and width of the filter (e.g., 3×3).

---

### Q75: Why do we commonly use 3×3 kernels?

Smallest kernel with spatial context, parameter efficient, stackable for larger receptive fields, proven performance.

---

### Q76: What is a stride?

Number of pixels the filter moves each step. Stride 1 = every position, stride 2 = skip every other.

---

### Q77: What is padding?

Adding extra pixels (usually zeros) around input borders to control output dimensions.

---

### Q78: What is "same" padding?

Adds enough padding so output size equals input size.

---

### Q79: What is "valid" padding?

No padding — filter only operates where it fully fits. Output size = input - kernel + 1.

---

### Q80: What happens to the image dimensions after convolution?

output_size = (input_size - kernel_size + 2*padding) / stride + 1

Our model: 28×28 → 26×26 (Conv1, valid, stride 1)

---

### Q81: What happens if stride increases?

Output size decreases, less overlap, downsampling effect, less computation.

---

### Q82: What happens if we use a larger kernel?

More parameters, larger receptive field, higher computational cost, may miss fine details.

---

### Q83: What is a convolutional layer learning?

A set of filters that detect specific patterns, transforming input into feature maps.

---

### Q84: What does the first convolution usually learn?

Low-level features: edges (horizontal, vertical, diagonal), lines, simple gradients.

---

### Q85: What might deeper convolutional layers learn?

Increasingly abstract features: textures, corners, shapes, object parts, complex patterns.

Parameters = (K_h × K_w × C_in + 1) × C_out
```

Where:
- K_h = 3 (kernel height)
- K_w = 3 (kernel width)
- C_in = 1 (input channels)
- C_out = 16 (number of filters)
- +1 for bias per filter

Calculation:
```
(3 × 3 × 1 + 1) × 16 = (9 + 1) × 16 = 10 × 16 = 160
```

---

### Q102: Calculate the parameters of your second convolution.

**4,640 parameters**

Formula:
```
Parameters = (K_h × K_w × C_in + 1) × C_out
```

Where:
- K_h = 3, K_w = 3
- C_in = 16 (output channels from Conv1)
- C_out = 32 (number of filters)
- +1 for bias per filter

Calculation:
```
(3 × 3 × 16 + 1) × 32 = (144 + 1) × 32 = 145 × 32 = 4,640
```

---

### Q103: How many parameters are in your Dense layer?

**801 parameters**

Formula:
```
Parameters = (N_inputs + 1) × N_outputs
```

Where:
- N_inputs = 800 (flattened features: 5 × 5 × 32)
- N_outputs = 1 (single neuron)
- +1 for bias

Calculation:
```
(800 + 1) × 1 = 801
```

---

### Q104: Why do we add 1?

The **+1** accounts for the **bias term**. Each filter/neuron has one bias value that shifts its activation.

---

### Q105: How many neurons enter your final Dense layer?

**800 neurons** enter the final Dense layer.

This comes from the Flatten layer: 5 × 5 × 32 = 800.

---

### Q106: How did you determine that number?

I traced the dimensions through the network:

| Layer | Input Shape | Output Shape |
|-------|-------------|--------------|
| Input | 28×28×1 | - |
| Conv1 | 28×28×1 | 26×26×16 |
| Pool1 | 26×26×16 | 13×13×16 |
| Conv2 | 13×13×16 | 11×11×32 |
| Pool2 | 11×11×32 | 5×5×32 |
| Flatten | 5×5×32 | 800 |

Flatten output = 5 × 5 × 32 = 800

---

### Q107: What is the output shape after every layer?

| Layer | Output Shape |
|-------|-------------|
| Input | (None, 28, 28, 1) |
| Conv1 | (None, 26, 26, 16) |
| Pool1 | (None, 13, 13, 16) |
| Conv2 | (None, 11, 11, 32) |
| Pool2 | (None, 5, 5, 32) |
| Flatten | (None, 800) |
| Dense | (None, 1) |

---

### Q108: Trace the dimensions from 28×28×1 through your entire network.

```

```

---

### Q109: What is the feature-map size after the first pooling layer?

**13 × 13 × 16**

Pool1 reduces spatial dimensions from 26×26 to 13×13 (factor of 2), keeping all 16 channels.

---

### Q110: What is the feature-map size after the second pooling layer?

**5 × 5 × 32**

Pool2 reduces from 11×11 to 5×5 (floor of 11/2), keeping all 32 channels.

---

### Q111: How many values enter Flatten?

**800 values**

Calculation: 5 (height) × 5 (width) × 32 (channels) = 800

---

### Q112: How does changing the number of filters affect parameter count?

More filters = more parameters in both that layer and the next layer.

Example:
- Conv1 with 16 filters: 160 parameters
- Conv1 with 32 filters: (3×3×1+1) × 32 = 320 parameters
- Conv2 input channels also change, affecting its parameter count

---

### Q113: What happens to the parameter count if you double the filters?

If we double filters from (16, 32) to (32, 64):

| Layer | Original | Doubled |
|-------|----------|---------|
| Conv1 | 160 | 320 |
| Conv2 | 4,640 | (3×3×32+1) × 64 = 18,496 |
| Dense | 801 | (5×5×64+1) × 1 = 1,601 |
| **Total** | **5,601** | **20,417** |

Doubling filters increases total parameters by ~3.6×.

---

### Q114: Why does the Dense layer potentially contain many parameters?

The Dense layer connects **every flattened feature to every output neuron**. With large feature maps, this creates many connections.

In our model:
- 800 features × 1 neuron = 800 weights + 1 bias = 801 parameters

If we had more filters or larger feature maps, the Dense layer would dominate the parameter count.

---

# 🔴 LEVEL 7 — ACTIVATION FUNCTIONS

---

### Q115: What is an activation function?

An activation function is a **non-linear function** applied to the output of a neuron/layer. It determines whether and how strongly the neuron "fires."

Without activation functions, a neural network would just be a linear transformation, unable to learn complex patterns.

---

### Q116: Why do neural networks need activation functions?

Neural networks need activation functions to **introduce non-linearity**. Without them, stacking layers would just be equivalent to a single linear transformation, regardless of depth.

Non-linearity allows the network to learn complex, non-linear relationships in data.

---

### Q117: What is ReLU?

ReLU (Rectified Linear Unit) is defined as:

```
ReLU(x) = max(0, x)
```

| Input x | ReLU(x) |
|---------|---------|
| -5 | 0 |
| -1 | 0 |
| 0 | 0 |
| 1 | 1 |
| 5 | 5 |

---

### Q118: Why use ReLU in the convolutional layers?

ReLU is used because:
1. **Prevents vanishing gradients** — gradient is 1 for positive values
2. **Computationally efficient** — simple max operation
3. **Sparsity** — many neurons output 0, improving efficiency
4. **Faster convergence** — compared to sigmoid/tanh

---

### Q119: What happens when ReLU receives a negative value?

ReLU outputs **0** for any negative input. This is called the "dying ReLU" problem when many neurons permanently output 0.

---

### Q120: Why not use sigmoid everywhere?

Sigmoid has problems:
1. **Vanishing gradients** — gradient approaches 0 for very large/small inputs
2. **Output not zero-centered** — outputs are all positive
3. **Computationally expensive** — exponential calculation
4. **Slow convergence** — compared to ReLU

---

### Q121: Why use sigmoid in the final layer?

Sigmoid is used in the final layer because:
1. **Output range (0, 1)** — perfect for probability
2. **Binary classification** — single output probability
3. **Standard practice** — sigmoid + binary cross-entropy is the standard combination

---

### Q122: What is the sigmoid formula?

```
σ(x) = 1 / (1 + e^(-x))
```

---

### Q123: What happens to sigmoid when x becomes very large?

When x → +∞:
```
σ(x) → 1
```

The output approaches 1 (model becomes very confident in positive class).

---

### Q124: What happens when x becomes very negative?

When x → -∞:
```
σ(x) → 0
```

The output approaches 0 (model becomes very confident in negative class).

---

# 🔴 LEVEL 8 — LOSS FUNCTIONS

---

### Q125: What loss function did you use?

**Binary cross-entropy** (binary_crossentropy in Keras).

---

### Q126: Why binary cross-entropy?

Binary cross-entropy is the standard loss for binary classification:
1. Mathematically appropriate for sigmoid output
2. Penalizes confident wrong predictions heavily
3. Smooth gradients for optimization
4. Directly measures the difference between predicted and actual probability distributions

---

### Q127: What is binary cross-entropy?

```
Loss = -[y·log(p) + (1-y)·log(1-p)]
```

Where:
- y = true label (0 or 1)
- p = predicted probability

For y=1: Loss = -log(p)
For y=0: Loss = -log(1-p)

---

### Q128: Why wouldn't we use mean squared error for this classification task?

MSE has problems for classification:
1. **Non-convex loss landscape** with sigmoid — many local minima
2. **Vanishing gradients** — when prediction is very wrong (p≈0 for y=1), gradient is tiny
3. **Slower convergence** — binary cross-entropy converges faster
4. **No probabilistic interpretation** — doesn't measure probability distribution difference

---

### Q129: What does the loss tell us?

The loss tells us **how far the model's predictions are from the true labels**. Lower loss = better predictions.

---

### Q130: Does lower loss always mean a better model?

**Not necessarily.** Lower training loss could indicate overfitting. We need to check validation loss too.

Example:
- Training loss 0.01, validation loss 0.5 → Overfitted
- Training loss 0.10, validation loss 0.12 → Better generalization

---

### Q131: What happens to the loss when the model makes a confident wrong prediction?

The loss becomes **very large**.

Example (y=1, model predicts p=0.01):
```
Loss = -log(0.01) = 4.605
```

Example (y=0, model predicts p=0.99):
```
Loss = -log(1-0.99) = -log(0.01) = 4.605

new_weight = old_weight - learning_rate × gradient

### Q220: Explain the complete forward pass of your CNN mathematically.

**Step 1: Conv1**
```
Input: X ∈ R^(28×28×1)
Conv1 filters: W1 ∈ R^(3×3×1×16), b1 ∈ R^16
Output1[i,j,k] = ReLU( Σ_{m,n,c} X[i+m, j+n, c] × W1[m,n,c,k] + b1[k] )
Output1: 26×26×16
```

**Step 2: Pool1**
```
Pool1[i,j,k] = max( Output1[2i:2i+2, 2j:2j+2, k] )
Pool1: 13×13×16
```

**Step 3: Conv2**
```
Conv2 filters: W2 ∈ R^(3×3×16×32), b2 ∈ R^32
Output2[i,j,k] = ReLU( Σ_{m,n,c} Pool1[i+m, j+n, c] × W2[m,n,c,k] + b2[k] )
Output2: 11×11×32
```

**Step 4: Pool2**
```
Pool2[i,j,k] = max( Output2[2i:2i+2, 2j:2j+2, k] )
Pool2: 5×5×32
```

**Step 5: Flatten**
```
Flatten = Pool2.reshape(800)
```

**Step 6: Dense**
```
z = W_dense · Flatten + b_dense
Output = σ(z) = 1 / (1 + e^(-z))
```

---

### Q221: Explain backpropagation through your CNN.

Backpropagation computes gradients using the chain rule:

1. **Compute loss gradient**: ∂L/∂Output

2. **Dense layer gradients**:
```
∂L/∂W_dense = ∂L/∂Output × Flatten^T
∂L/∂Flatten = W_dense^T × ∂L/∂Output
```

3. **Flatten gradient** (reshape to 5×5×32):
```
∂L/∂Pool2 = ∂L/∂Flatten.reshape(5, 5, 32)
```

4. **Pool2 gradient** (route through max positions):
```
∂L/∂Output2[i,j,k] = ∂L/∂Pool2[i//2, j//2, k] if Output2[i,j,k] was max
```

5. **Conv2 gradients** (convolution with flipped filters):
```
∂L/∂W2 = Pool1 ⊗ ∂L/∂Output2
∂L/∂Pool1 = full_convolution(∂L/∂Output2, W2)
```

6. **Continue back through Pool1 and Conv1 similarly**

7. **Update weights**: W = W - η × ∂L/∂W

---

### Q222: How are convolutional filter weights updated?

Filter weights are updated via gradient descent:

```
W_new = W_old - learning_rate × ∂L/∂W
```

Where ∂L/∂W is computed by convolving the input with the error gradient:

```
∂L/∂W = Input ⊗ Error_Gradient
```

This updates all filter weights simultaneously based on their contribution to the loss.

---

### Q223: How does the chain rule appear in backpropagation?

The chain rule appears throughout backpropagation:

```
∂L/∂W1 = ∂L/∂Output1 × ∂Output1/∂W1
       = ∂L/∂Pool1 × ∂Pool1/∂Output1 × ∂Output1/∂W1
       = ∂L/∂Conv2 × ∂Conv2/∂Pool1 × ∂Pool1/∂Output1 × ∂Output1/∂W1
```

Each layer's gradient depends on gradients from all subsequent layers.

---

### Q224: Why can deep networks suffer from vanishing gradients?

Vanishing gradients occur when **gradients become extremely small** as they propagate backward through many layers.

Cause:
1. Activation functions with small derivatives (sigmoid derivative max = 0.25)
2. Repeated multiplication of small values
3. After many layers, gradient ≈ 0
4. Early layers stop learning

### Q225: How does ReLU help with gradient flow?

ReLU helps because:
1. **Derivative is 1** for positive inputs
2. **No shrinking** — gradient passes through unchanged

ReLU derivative:

### Q235: Why is binary cross-entropy mathematically appropriate for sigmoid binary classification?

Binary cross-entropy is appropriate because:
1. **Maximum likelihood** — minimizes negative log-likelihood
2. **Gradient is well-behaved** — ∂L/∂z = p - y (difference between prediction and truth)
3. **Natural pairing** — sigmoid output + BCE loss produces clean gradients
4. **Convex optimization** — for linear models, loss surface is convex

Gradient of BCE with sigmoid:
```
∂L/∂z = σ(z) - y
```
This simple gradient prevents vanishing gradient problems.
